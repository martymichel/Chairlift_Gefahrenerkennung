```python
import cv2
import numpy as np
import time
import tkinter as tk
from tkinter import ttk
from PIL import Image, ImageTk

class VideoPlayer:
    def __init__(self, video_source):
        self.cap = cv2.VideoCapture(video_source)
        if not self.cap.isOpened():
            raise ValueError("Unable to open video source", video_source)

        self.history = 500
        self.dist2Threshold = 100
        self.running = True  # Flag für den Videoplayer-Status

        self.fgbg = cv2.createBackgroundSubtractorKNN(
            history=self.history, dist2Threshold=self.dist2Threshold, detectShadows=False)

        self.original_fps = self.cap.get(cv2.CAP_PROP_FPS)
        self.frame_delay = int(1000 / self.original_fps)

        # Original Video-Dimensionen
        self.original_width = int(self.cap.get(cv2.CAP_PROP_FRAME_WIDTH))
        self.original_height = int(self.cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

        self.root = tk.Tk()
        self.root.title("Video Player")
        
        # Mindestfenstergröße setzen
        self.root.minsize(400, 300)
        
        # Event-Handler für das Schließen des Fensters
        self.root.protocol("WM_DELETE_WINDOW", self.on_closing)

        # History Slider - oben im Fenster
        self.history_label = tk.Label(self.root, text="History")
        self.history_label.pack()
        self.history_slider = ttk.Scale(self.root, from_=1, to=1000, orient=tk.HORIZONTAL,
                                        command=self.update_history)
        self.history_slider.set(self.history)
        self.history_slider.pack(fill=tk.X)

        # Distance Threshold Slider - oben im Fenster (korrigierter Name für KNN)
        self.threshold_label = tk.Label(self.root, text="Distance Threshold")
        self.threshold_label.pack()
        self.dist2Threshold_slider = ttk.Scale(self.root, from_=1, to=500, orient=tk.HORIZONTAL,
                                             command=self.update_dist2Threshold)
        self.dist2Threshold_slider.set(self.dist2Threshold)
        self.dist2Threshold_slider.pack(fill=tk.X)

        # Canvas - expandiert mit dem Fenster
        self.canvas = tk.Canvas(self.root, bg='black')
        self.canvas.pack(fill=tk.BOTH, expand=True)
        
        # Event-Handler für Fenstergrößenänderungen
        self.canvas.bind('<Configure>', self.on_canvas_configure)
        
        # Aktuelle Canvas-Dimensionen
        self.canvas_width = 400
        self.canvas_height = 300

        # Video-Update erst nach der vollständigen Initialisierung starten
        self.root.after(100, self.update)

        self.root.mainloop()

    def on_closing(self):
        """Wird aufgerufen, wenn das Fenster geschlossen wird"""
        self.running = False
        self.cleanup()
        self.root.destroy()

    def cleanup(self):
        """Räumt Ressourcen auf"""
        if self.cap:
            self.cap.release()
        print("Video Player beendet und Ressourcen freigegeben.")

    def on_canvas_configure(self, event):
        """Wird aufgerufen, wenn sich die Canvas-Größe ändert"""
        self.canvas_width = event.width
        self.canvas_height = event.height

    def calculate_display_size(self):
        """Berechnet die optimale Anzeigegröße basierend auf der Canvas-Größe"""
        # Verfügbare Canvas-Größe (mal 2 wegen gestapeltem Bild)
        available_width = self.canvas_width
        available_height = self.canvas_height
        
        # Original-Seitenverhältnis des gestapelten Videos (2x Höhe)
        original_aspect = self.original_width / (self.original_height * 2)
        
        # Berechne Skalierung basierend auf verfügbarem Platz
        scale_w = available_width / self.original_width
        scale_h = available_height / (self.original_height * 2)
        
        # Verwende die kleinere Skalierung, um das Seitenverhältnis zu bewahren
        scale = min(scale_w, scale_h, 1.0)  # Maximal Originalgröße
        
        display_width = int(self.original_width * scale)
        display_height = int(self.original_height * scale)
        
        return display_width, display_height

    def update_history(self, value):
        self.history = int(float(value))
        self.update_bg_subtractor()

    def update_dist2Threshold(self, value):
        self.dist2Threshold = int(float(value))
        self.update_bg_subtractor()

    def update_bg_subtractor(self):
        self.fgbg = cv2.createBackgroundSubtractorKNN(
            history=self.history, dist2Threshold=self.dist2Threshold, detectShadows=False)

    def update(self):
        if not self.running:
            return  # Stoppe die Updates, wenn das Fenster geschlossen wird
        
        # Überprüfen ob Canvas existiert und bereit ist
        try:
            if not self.canvas.winfo_exists():
                return
        except tk.TclError:
            return
            
        ret, frame = self.cap.read()
        if ret:
            fgmask = self.fgbg.apply(frame)
            kernel = np.ones((5, 5), np.uint8)
            fgmask = cv2.morphologyEx(fgmask, cv2.MORPH_OPEN, kernel)

            contours, _ = cv2.findContours(fgmask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
            for contour in contours:
                if cv2.contourArea(contour) < 500:
                    continue
                (x, y, w, h) = cv2.boundingRect(contour)
                cv2.rectangle(frame, (x, y), (x + w, y + h), (0, 255, 0), 2)

            # Frames stapeln
            stacked_frame = np.vstack((frame, cv2.cvtColor(fgmask, cv2.COLOR_GRAY2BGR)))
            
            # Dynamische Größenberechnung basierend auf aktueller Canvas-Größe
            display_width, display_height = self.calculate_display_size()
            
            # Frame skalieren für die Anzeige
            stacked_frame_resized = cv2.resize(stacked_frame, (display_width, display_height*2))
            stacked_frame_resized = cv2.cvtColor(stacked_frame_resized, cv2.COLOR_BGR2RGB)
            
            try:
                img = ImageTk.PhotoImage(image=Image.fromarray(stacked_frame_resized))

                # Canvas leeren und zentriert anzeigen
                self.canvas.delete("all")
                x_center = self.canvas_width // 2
                y_center = self.canvas_height // 2
                self.canvas.create_image(x_center, y_center, image=img)
                self.canvas.image = img
            except tk.TclError as e:
                print(f"Tkinter Fehler beim Anzeigen des Bildes: {e}")
        else:
            self.cap.set(cv2.CAP_PROP_POS_FRAMES, 0)
            ret, frame = self.cap.read()

        if self.running:  # Nur weitermachen, wenn der Player noch läuft
            self.root.after(self.frame_delay, self.update)

# Ersetze 'your_video.mp4' mit dem Pfad zu deinem Video
VideoPlayer(loadVideoPath())
```

```python
import cv2
import numpy as np

cap = loadVideo()

#  Standardparameter
history = 500
varThreshold = 16
algo = 'MOG2'

# Hintergrundsubtraktor initialisieren
def initialize_subtractor(algo, history, varThreshold):
    if algo == 'MOG2':
        return cv2.createBackgroundSubtractorMOG2(history=history, varThreshold=varThreshold, detectShadows=False)
    elif algo == 'KNN':
        return cv2.createBackgroundSubtractorKNN(history=history, dist2Threshold=varThreshold, detectShadows=False)
    else:
        return cv2.createBackgroundSubtractorMOG2(history=history, nmixtures=3, backgroundRatio=0.7, noiseSigma=0)

fgbg = initialize_subtractor(algo, history, varThreshold)

# Callback-Funktion für die Schieberegler
def on_history_change(val):
    global history, fgbg
    history = val
    fgbg = initialize_subtractor(algo, history, varThreshold)

def on_threshold_change(val):
    global varThreshold, fgbg
    varThreshold = val
    fgbg = initialize_subtractor(algo, history, varThreshold)

# Hauptfenster erstellen und als vergrößerbar/verkleinerbar konfigurieren
window_name = 'Frame'
cv2.namedWindow(window_name, cv2.WINDOW_NORMAL)
cv2.createTrackbar('History', window_name, history, 1000, on_history_change)
cv2.createTrackbar('Threshold', window_name, varThreshold, 255, on_threshold_change)

print("Video Player gestartet. Steuerung:")
print("- Taste '1': Wechsel zu KNN Algorithmus")
print("- Taste '2': Wechsel zu MOG2 Algorithmus") 
print("- Taste 'q': Beenden")
print("- Fenster schließen: Beenden")

# Flag um zu verfolgen ob das Programm läuft
running = True
window_was_visible = True

try:
    while running:
        ret, frame = cap.read()
        if not ret:
            # Wenn das Video zu Ende ist, setzen Sie es zurück
            cap.set(cv2.CAP_PROP_POS_FRAMES, 0)
            continue

        # Hintergrundsubtraktion anwenden
        fgmask = fgbg.apply(frame)

        # Morphologische Operationen zur Verbesserung der Maske
        kernel = np.ones((5, 5), np.uint8)
        fgmask = cv2.morphologyEx(fgmask, cv2.MORPH_OPEN, kernel)

        # Konturen finden
        contours, _ = cv2.findContours(fgmask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

        # Zeichnen der Konturen
        for contour in contours:
            if cv2.contourArea(contour) < 500:  # Ignorieren kleiner Konturen
                continue
            (x, y, w, h) = cv2.boundingRect(contour)
            cv2.rectangle(frame, (x, y), (x + w, y + h), (0, 255, 0), 2)

        # Überprüfen ob das Fenster noch existiert (vor imshow)
        try:
            window_prop = cv2.getWindowProperty(window_name, cv2.WND_PROP_VISIBLE)
            if window_prop < 0:  # Fenster existiert nicht mehr
                if window_was_visible:  # Nur beim ersten Mal ausgeben
                    print("Fenster wurde geschlossen. Beende Programm...")
                    window_was_visible = False
                running = False
                break
        except cv2.error:
            if window_was_visible:
                print("Fenster wurde geschlossen. Beende Programm...")
                window_was_visible = False
            running = False
            break

        # Nur anzeigen wenn das Fenster noch existiert
        if running:
            try:
                cv2.imshow(window_name, frame)
            except cv2.error:
                print("Fehler beim Anzeigen. Fenster wurde geschlossen...")
                running = False
                break

        # Tastatursteuerung für Algorithmusauswahl (nur wenn Fenster noch offen)
        if running:
            key = cv2.waitKey(1) & 0xFF
            if key == ord('1'):
                algo = 'KNN'
                fgbg = initialize_subtractor(algo, history, varThreshold)
                print(f"Algorithmus gewechselt zu: {algo}")
            elif key == ord('2'):
                algo = 'MOG2'
                fgbg = initialize_subtractor(algo, history, varThreshold)
                print(f"Algorithmus gewechselt zu: {algo}")
            elif key == ord('q'):
                print("Beende auf Benutzerwunsch...")
                running = False
                break
            elif key == 27:  # ESC Taste
                print("Beende durch ESC-Taste...")
                running = False
                break

except KeyboardInterrupt:
    print("Programm durch Benutzer unterbrochen...")
    running = False
except Exception as e:
    print(f"Fehler aufgetreten: {e}")
    running = False
finally:
    # Aufräumen - wird immer ausgeführt
    print("Räume Ressourcen auf...")
    try:
        cap.release()
    except:
        pass
    try:
        cv2.destroyAllWindows()
    except:
        pass
    print("Video Player beendet.")
```
